```python
def right_triangle(x, y, z):
    if x*x + y*y == z*z:
        return True
    elif x*x + z*z == y*y:
        return True
    elif y*y + z*z == x*x:
        return True
    else:
        return False


try:
    x = float(input("Enter the first side of triangle: "))
    y = float(input("Enter the second side of triangle: "))
    z = float(input("Enter the third side of triangle: "))

    if right_triangle(x, y, z):
        print("It is right angled triangle")
    else:
        print("It is not a right angled triangle")

except ValueError:
    print("Invalid input")
```
