```python
#Q9

from datetime import date
d1=int(input('Day 1:'))
m1=int(input("Month 1:"))
y1=int(input("Year 1"))
d2=int(input('Day 2:'))
m2=int(input("Month 2:"))
y2=int(input("Year 2:"))

date1=date(y1,m1,d1)
date2=date(y2,m2,d2)

days=abs((date2-date1).days)
print("Days between dates:",days)
```
